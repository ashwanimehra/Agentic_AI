```python
from langchain.agents import create_agent
```

```python
import random
def get_weather(city:str)->str:
    """returns the weather in the city"""
    weathers = ["sunny", "cloudy", "rainy", "cold"]
    return f"The weather in {city} is {random.choice(weathers)}!"
```

```python
agent = create_agent(
    model="google_genai:gemini-3-flash-preview",
    tools=[get_weather],
    system_prompt="you are a helpful assistant"
)
```

```python
agent
```

```python
response = agent.invoke({
    "messages":[
        {"role":"user","content":"What is the weather in Gurugram?"}
        ]
        })
```

```python
response
```

```python
response["messages"][-1].text
```

# init_chat_model

```python
from langchain.chat_models import init_chat_model
```

```python
model = init_chat_model(
    "google_genai:gemini-3-flash-preview"
)
```

```python
model
```

```python
model_with_tools = model.bind_tools([get_weather])
```

```python
model_with_tools
```

# Streaming

```python
for chunk in model.stream("what is Reasoning? and what topics does it contains??"):
    print(chunk.text, flush=True)
```

# Batching

```python
query_inputs = [
    {"messages":[{"role":"user", "content":"what is a Lunar year?"}]},
    {"messages":[{"role":"user", "content":"what is batching in Agentic AI?"}]},
    {"messages":[{"role":"user", "content":"whats 2+2?"}]},
    {"messages":[{"role":"user", "content":"what is the weather in Rewari, gurugram and Delhi?"}]},
    ]

agent_batch_stream = agent.batch_as_completed(
    inputs=query_inputs,
    config={
        "max_concurrency":2
    }
)

for chunk in agent_batch_stream:
    print(chunk, flush=True)
```

```python
model.batch(
    inputs=[
        "what is a Lunar year?",
        "what is batching in Agentic AI?",
        "whats 2+2?",
        "what is the weather in Rewari, gurugram and Delhi?"
    ],
    config={
        "max_concurrency":2
    }
)
```
